```python
#How many times were customers contacted?
df['campaign'].value_counts().sort_index()
sns.countplot(data=df, x='campaign')
plt.title('Number of Contacts per Customer')
plt.xlabel('Number of Contacts')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#compare successful and unsuccessful marketing campaigns

df['y'].value_counts() #Count successful and unsuccessful outcomes
```

```python
df['y'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Campaign Success vs Failure')
plt.ylabel('')
plt.show()
```

```python
#Examine the relationship between previous campaign results and current subscriptions.

df['poutcome'].value_counts()
pd.crosstab(df['poutcome'], df['y'])
pd.crosstab(df['poutcome'], df['y'], normalize='index') * 100
sns.countplot(data=df, x='poutcome', hue='y')

plt.title('Previous Campaign Result vs Current Subscription')
plt.show()
```

```python
#Identify customer segments and campaign patterns associated with higher subscription rates.
#Subscription by job
pd.crosstab(df['job'], df['y'], normalize='index') * 100
```

```python
sns.barplot(data=df, x='job', y=df['y'].eq('yes').astype(int))

plt.xticks(rotation=45)
plt.title('Subscription by Job')
plt.show()
```

```python
#Subscription rate by age group
pd.crosstab(df['age_group'], df['y'], normalize='index') * 100

age = pd.crosstab(df['age_group'], df['y'], normalize='index') * 100

age.plot(kind='bar', stacked=True)

plt.title('Subscription Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Percentage (%)')
plt.show()
```

```python
#Subscription by education
pd.crosstab(df['education'], df['y'], normalize='index') * 100
```

```python
education = pd.crosstab(df['education'], df['y'], normalize='index') * 100

education.plot(kind='bar', stacked=True)

plt.title('Subscription Rate by Education')
plt.xlabel('Education')
plt.ylabel('Percentage (%)')
plt.show()
```

```python
#Subscription rate by campaign month

pd.crosstab(df['month'], df['y'], normalize='index') * 100
```

```python

month = pd.crosstab(df['month'], df['y'], normalize='index') * 100

month.plot(kind='bar', stacked=True)

plt.title('Subscription Rate by Campaign Month')
plt.xlabel('Month')
plt.ylabel('Percentage (%)')
plt.show()
```
